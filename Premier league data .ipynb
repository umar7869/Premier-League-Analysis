```python



teams = {}

def get_team(name):
    """Return a team's stats dict, creating it if it doesn't exist yet"""
    if name not in teams:
        teams[name] = {'Team': name, 'Played': 0, 'Wins': 0, 'Draws': 0,
                        'Losses': 0, 'GF': 0, 'GA': 0, 'Points': 0,'Shots':0,'Shots on Target':0,'Corners':0}
    return teams[name]

# Loop through every match and update both teams involved
for _, row in df.iterrows():
    home = get_team(row['HomeTeam'])
    away = get_team(row['AwayTeam'])

    home['Played'] += 1
    away['Played'] += 1
    
    home['GF'] += row['FTHG']
    home['GA'] += row['FTAG']
    away['GF'] += row['FTAG']
    away['GA'] += row['FTHG']
        
    home['Shots']+=row['HS']
    away['Shots']+=row['AS']
    
    home['Shots on Target']+=row['HST']
    away['Shots on Target']+=row['AST']

    home['Corners']+=row['HC']
    away['Corners']+=row['AC']

    

    

    if row['FTR'] == 'H':          # home team won
        home['Wins'] += 1; home['Points'] += 3
        away['Losses'] += 1
    elif row['FTR'] == 'A':        # away team won
        away['Wins'] += 1; away['Points'] += 3
        home['Losses'] += 1
    else:                          # draw
        home['Draws'] += 1; home['Points'] += 1
        away['Draws'] += 1; away['Points'] += 1


table = pd.DataFrame(teams.values())
table['GD'] = table['GF'] - table['GA']


table = table.sort_values(['Points', 'GD'], ascending=False).reset_index(drop=True)
table.index += 1 

print(table)
```
